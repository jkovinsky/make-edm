```python
import csv, json
from datetime import datetime, date
```

```python
with open('2026-08-07_friday.csv', 'r') as f:
    reader = csv.reader(f)
    data = list(reader)
```

```python
data[:5]
```

```python
time_now = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
```

```python
time_now
```

```python
out = []
for row in data:
    map = [{'artists': row, 'date': time_now}]
    out.append(map)
```

```python
with open('outside_lands_friday.json', 'w') as f:
    json.dump(out, f, indent=4)
```
