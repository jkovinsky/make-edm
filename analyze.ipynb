```python
import json
```

```python
with open('spotify_results.json', 'r') as f:
    data = json.load(f)
```

```python
scores = [item['popularity'] for item in data]
```

```python
spread = {}
for score in scores:
    if score not in spread:
        spread[score] = 1
    else:
        spread[score] += 1
```

```python
sorted_dict = {k: v for k, v in sorted(spread.items(), key=lambda item: item[1])}
```

```python
sorted_dict
```

```python
seen = {}
for a in data:
    if a['id'] not in seen:
        seen[a['id']] = a
data = list(seen.values())
```

```python
import random

take = {(8, 10): 4, (5, 7): 11, (3, 4): 15}

playlist = []
for (low, high), n in take.items():
    tier = [a for a in data if a['popularity'] and low <= a['popularity'] <= high]
    playlist.extend(random.sample(tier, min(n, len(tier))))
```
